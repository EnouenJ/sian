```python
import numpy as np
import matplotlib.pyplot as plt
import torch
import torch.nn as nn

from sian.synth.causal_synthetic.HyperPFN import HDAG
from sian.synth.causal_synthetic.HyperPFN import Doable_HDAG_SEM

%load_ext autoreload
%autoreload 2
```

```python
default_dist_info = {'type' : 'log_uniform', 'low' : 0.5, 'high' : 2.0}

ERDOS_OR_BARABOSI = 'barabosi'
ERDOS_OR_BARABOSI = 'erdos'
if ERDOS_OR_BARABOSI=='barabosi':
    D = 10
    m_list = [2]
    graph_gen_type = "barabosi_hyper"
elif ERDOS_OR_BARABOSI=='erdos':
    D = 10
    m_list = [2*D, 1*D]
    graph_gen_type = "erdos_hyper"


full_meta_hyperparameters = {
    "graph_generation" : {
        "D" : D,
        "graph_gen_type" : graph_gen_type,
        "m_list" : m_list,
    },

    "SEM_generation" : {
        "cts_dic_ratio" : 0.5,
        "min_max_discrete_categories" : [2,10],
    
        "cts_function_options_list" : [{'function_type':'multilinear','coef_dist_info':default_dist_info}],
        "disc_function_options_list" : [{'function_type':'multilinear','coef_dist_info':default_dist_info}],
        "cts_noise_options_list" : [{'noise_type':'gaussian','std_dist_info':default_dist_info}, ],
        "disc_noise_options_list" : [{'noise_type':'disc.gumbel'}, ],
    }
}
```

```python
myhdag = Doable_HDAG_SEM(full_meta_hyperparameters)
```

```python
N=10*1000
X_dict = myhdag.generate_data(N)
```

```python
for j in X_dict:
    X_j = X_dict[j]

    if X_j.shape[1]==1:
        sig_j = np.var(X_j)
        print(j,'  var',sig_j)
    else:
        Ik=X_j.shape[1]
        p_j = np.mean(X_j,axis=0)
        # print('p_j',p_j)
        ent_j = - np.sum( p_j * np.log(p_j) )
        ent2_j = ( np.log(Ik) ) - ( ent_j )
        # print(j,ent2_j)
        print(j,'  ent',ent_j)

```
